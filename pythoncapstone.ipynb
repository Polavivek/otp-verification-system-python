```python
import random
import smtplib
import ssl
from email.message import EmailMessage

# Configuration
SENDER_EMAIL = "Polavivek18@gmail.com"        # Replace with your email
APP_PASSWORD = "stwscsbssdcsokno"      # Replace with your app password
SMTP_SERVER = "smtp.gmail.com"
SMTP_PORT = 465

# 1. Generate OTP
def generate_otp(length=6):
    """Generates a random 6-digit OTP."""
    return ''.join([str(random.randint(0, 9)) for _ in range(length)])

# 2. Send OTP via Email
def send_otp_email(receiver_email, otp):
    """Sends the generated OTP to the user's email address."""
    try:
        msg = EmailMessage()
        msg.set_content(f"Your OTP is: {otp}")
        msg['Subject'] = "Your One-Time Password (OTP)"
        msg['From'] = SENDER_EMAIL
        msg['To'] = receiver_email

        context = ssl.create_default_context()
        with smtplib.SMTP_SSL(SMTP_SERVER, SMTP_PORT, context=context) as server:
            server.login(SENDER_EMAIL, APP_PASSWORD)
            server.send_message(msg)

        print(f"✅ OTP sent to {receiver_email}.")
    except Exception as e:
        print("❌ Failed to send OTP:", e)

# 3. Get OTP Input from User
def get_user_input():
    """Prompts user to input the OTP."""
    return input("Enter the OTP sent to your email: ").strip()

# 4. Verify OTP
def verify_otp(generated_otp, entered_otp):
    """Checks if the entered OTP matches the generated OTP."""
    return generated_otp == entered_otp

# 5. Main Program Flow
def run_otp_verification():
    print("\n--- OTP Verification System ---")
    user_email = input("Enter your email address: ").strip()

    otp = generate_otp()
    send_otp_email(user_email, otp)

    max_attempts = 3
    for attempt in range(max_attempts):
        user_otp = get_user_input()
        if verify_otp(otp, user_otp):
            print("✅ OTP verified successfully. Access granted!")
            return
        else:
            print(f"❌ Incorrect OTP. Attempts left: {max_attempts - attempt - 1}")

    print("🚫 Access denied. Too many failed attempts.")

# Run the program
if __name__ == "__main__":
    run_otp_verification()
```
